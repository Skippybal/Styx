```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
data = [np.clip(np.random.normal(0.5, 0.2, 1000), 0, 1)] * 10

# Create the box plot
plt.boxplot(data, whis=(0,100))
# plt.boxplot(data)

# Customize the plot (optional)
plt.ylim(bottom=-0.1, top=1.1)
plt.title("")
plt.xlabel("Config")
plt.ylabel("Utility")

# Show the plot
plt.show()
```

```python
means = [0.87, 0.3, 0.25, 0.5, 0.1, 0.05, 0.4, 0.32]
std = [0.04, 0.03, 0.06, 0.01, 0.01 , 0.05, 0.03, 0.02]
for i, mean in enumerate(means):
    data[i+2] = np.clip(np.random.normal(mean, std[i], 1000), 0, 1)
# Create the box plot
plt.boxplot(data, whis=(0, 100))
# plt.boxplot(data)

# Customize the plot (optional)
plt.ylim(bottom=-0.1, top=1.1)
plt.title("")
plt.xlabel("Config")
plt.ylabel("Utility")

# Show the plot
plt.show()
```
