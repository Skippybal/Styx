```python
smac_randr = pd.read_csv("storage_alice/state-run1/runs_and_results-it20.csv")
print(smac_randr.columns)
```

```python
print(len(smac_randr["Cutoff Time Used"]))
print(smac_randr["Cutoff Time Used"].value_counts())
# print(len(smac_randr["Cutoff Time Used"]))
```

```python
grouped = smac_randr.groupby("Run History Configuration ID")
n_groups = 0
group_sizes = []
for idx, g in grouped:
    # print(len(g))
    # display(g)
    # print(len(g))
    group_sizes.append(len(g))
    n_groups += 1

print(f"total number of configs tried: {n_groups}")
print(f"avg_instance_per_config: {np.mean(group_sizes)}")
```

```python
import json

""

with open("./storage_alice/smac3_output/smac3_mipverify/0/runhistory.json") as filehandle:
    smac3_data = json.load(filehandle)

print(smac3_data)
```

```python
storage = np.zeros(len(smac3_data['configs']))
for run in smac3_data["data"]:
    storage[run["config_id"]-1] += 1

print(storage)
```

```python
print(f"total_runs: {smac3_data['stats']['finished']}")
print(f"total_runs at max captime: {smac3_data['stats']['finished']}")
print(f"total_configs tried: {len(smac3_data['configs'])}")
print(f"average instances per configuration: {np.mean(storage)}")
```
